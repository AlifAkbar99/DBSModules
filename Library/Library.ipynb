## Regex

Reguler expression adalah sebuah cara mencari teks dengan pola tertentu. Pada Regex kita mencari sebuah kata dengan pola yang diinginkan.

````python
import re

pola= '^a...s$'
string_tes = 'abyss'
hasil= re.match(pola, string_tes)

if hasil:
  print('pencarian berhasil')
else:
  print('tidak berhasil')
````

### 3. Matplotlib
library untuk melakukan visualisasi data

````python
import matplotlib.pyplot as plt

# Data
x = [1,2,3,4,5]
y = [2,4,6,8,10]

# Membuat plot garis
plt.plot(x, y)

# Menambahkan judul
plt.title("Contoh plot garis")
plt.xlabel("Sumbu x")
plt.ylabel("Sumbu y")

# Menampilkan grafik
plt.show()
````

## Web Scrapping
jenis library untuk membantu pengguna mengumpulkan data dari halaman web.

### 1. Beautifulsoup
 library untuk mengambil data dari halaman web dan mengekstrak informasi yang diperlukan.

 ```
 pip install beautifulsoup4
 ```

````python
from urllib.request import urlopen
from bs4 import BeautifulSoup

url = "https://python.org/"
page = urlopen(url)

html = page.read()  # jangan decode dulu
soup = BeautifulSoup(html, "html.parser")

print(soup.title)
````

### 2. Urllib
Library yang digunakan untuk scraping konten dari sebuah website.

````python
from urllib.request import urlopen
import gzip

url = "https://python.org/"
page = urlopen(url)

html = gzip.decompress(page.read()).decode("utf-8")

start_index = html.find("<title>") + len("<title>")
end_index = html.find("</title>")

title = html[start_index:end_index]
print(title)
````
